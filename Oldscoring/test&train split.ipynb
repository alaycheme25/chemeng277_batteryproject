```python
# Importing the new_data file
pdata = pd.read_excel('mp_battery_data_new.csv.xlsx')
```

```python
# Split the data into 80% training and 20% testing sets, randomly
# random_state ensures the split is reproducible every time you run the code
X_train, X_test = train_test_split(pdata, test_size=0.2, random_state=42, shuffle=True)
```

```python
print("Training set shape:", X_train.shape)
print("Test set shape:", X_test.shape)
```

```python
# You can now apply an unsupervised learning algorithm to the X_train data
# For example, fitting a K-Means clustering model
from sklearn.cluster import KMeans
from sklearn.model_selection import KFold
```

```python
# 2. Define the number of folds (n_splits=5 is common)
N_FOLDS = 5
kf = KFold(n_splits=N_FOLDS, shuffle=True, random_state=42)
```

```python
# 3. Initialize an empty list to store scores
max_delta_volume_scores = []
```

```python
for train_index, test_index in kf.split(pdata):
    X_train, X_test = pdata[train_index], pdata[test_index]

    # 5. Define the unsupervised model (e.g., K-Means with k=4 clusters)
    # The 'k' here is a hyperparameter we are evaluating
    model = KMeans(n_clusters=5, init='k-means++', n_init=10, random_state=42)

    # 6. Train the model on the training fold
    model.fit(X_train)

    # 7. Evaluate the model on the test fold using a relevant metric
    # For K-Means, the inertia_ (within-cluster sum of squares) is a common metric
    # Note: Metrics for unsupervised learning assess structure, not "accuracy"
    score = model.max_delta_volume
    max_delta_volume_scores_scores.append(score)
```
